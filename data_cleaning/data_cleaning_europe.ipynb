```python
import pandas as pd
from pathlib import Path

# ----------------------------
# CONFIG
# ----------------------------
BASE_DIR = Path(r"C:\Users\Sagar Kumar\Desktop\portfolio\data\europe")

START_DATE = pd.to_datetime("2016-04-01")
END_DATE = pd.to_datetime("2025-03-26")

# ----------------------------
# PROCESS EUROPE EXCEL FILES
# ----------------------------
excel_files = list(BASE_DIR.glob("*.xlsx")) + list(BASE_DIR.glob("*.xls"))

print("\n📂 Processing Europe Excel files")

for file_path in excel_files:
    print(f"▶ Processing: {file_path.name}")

    # Read Excel
    df = pd.read_excel(file_path)

    # Rename first column to 'date'
    df.rename(columns={df.columns[0]: "date"}, inplace=True)

    # Convert date column
    df["date"] = pd.to_datetime(df["date"], errors="coerce")

    # Drop invalid dates
    df = df.dropna(subset=["date"])

    # Filter required date range
    df = df[
        (df["date"] >= START_DATE) &
        (df["date"] <= END_DATE)
    ]

    # Sort ascending by date
    df = df.sort_values(by="date", ascending=True)

    # Remove time part (YYYY-MM-DD)
    df["date"] = df["date"].dt.date

    # Save as CSV with SAME name
    output_csv = file_path.with_suffix(".csv")
    df.to_csv(output_csv, index=False)

    # Delete original Excel file
    file_path.unlink()

print("\n✅ All Europe Excel files filtered, sorted, and saved as CSV successfully.")
```
